```python
import pandas as pd
import matplotlib.pyplot as plt
import pyBigWig
import seaborn as sns
import numpy as np
```

```python
circRNA_df = pd.read_csv("../../resources/circRNA subcellular localization information.txt",sep = '\t')
```

```python
circRNA_location = circRNA_df
```

```python
TITLE_SIZE = 16
LABEL_SIZE = 14
TICK_SIZE = 12
LEGEND_SIZE = 12
LINE_WIDTH = 1.5
BAR_ALPHA = 0.85

warm_colors = [
    '#F8D7B4', '#F0C9A8', '#E8BB99', '#E0AD8A', '#D89F7B', '#D0916C',
    '#C8835D', '#C0754E', '#B8673F', '#B05930', '#A84B21'
]
cool_colors = [
    '#D4E6F1', '#C5D9E8', '#B6CCDF', '#A7BFD6', '#98B2CD',
    '#89A5C4', '#7A98BB', '#6B8BB2', '#5C7EA9'
]

def beautify_ax(ax):
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.tick_params(axis='both', labelsize=TICK_SIZE)
    ax.grid(axis='y', linestyle=':', linewidth=0.7, alpha=0.5)
```

```python
# Plot the RNALocate_Score distribution.
print(len(circRNA_df))
plt.figure(figsize=(8, 6))
plt.hist(circRNA_df['RNALocate_Score'], bins=20, range=(0.3, 1.0), color='skyblue', edgecolor='black')

# Add plot title and axis labels.
plt.title('Distribution of RNALocate_Score', fontsize=14)
plt.xlabel('RNALocate_Score', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

# Show the plot.
plt.tight_layout()
plt.show()
```

```python
fig_a, ax_a = plt.subplots(figsize=(8, 6))
ax_a.hist(
    circRNA_df['RNALocate_Score'],
    bins=20, range=(0.3, 1.0),
    color=cool_colors[5], edgecolor='black', alpha=BAR_ALPHA
)
ax_a.axvline(0.5, color=warm_colors[8], linestyle='--', linewidth=LINE_WIDTH, label='Cutoff = 0.5')
ax_a.set_title('Distribution of RNALocate_Score', fontsize=TITLE_SIZE, fontweight='bold')
ax_a.set_xlabel('RNALocate Score', fontsize=LABEL_SIZE)
ax_a.set_ylabel('Frequency', fontsize=LABEL_SIZE)
ax_a.legend(frameon=False, fontsize=LEGEND_SIZE)
beautify_ax(ax_a)
plt.tight_layout()
```

```python
localization_counts = circRNA_df['Subcellular_Localization'].value_counts()

# Count each element type.
print(localization_counts)

# Draw the bar plot.
plt.figure(figsize=(10, 6))
localization_counts.plot(kind='bar', color='skyblue', edgecolor='black')

# Add plot title and axis labels.
plt.title('Distribution of Subcellular Localization', fontsize=14)
plt.xlabel('Subcellular Localization', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

# Show the plot.
plt.tight_layout()
plt.show()
```

```python
loc_counts_before = circRNA_df['Subcellular_Localization'].value_counts()
fig_b, ax_b = plt.subplots(figsize=(10, 6))
bars = ax_b.bar(
    loc_counts_before.index, loc_counts_before.values,
    color=cool_colors[4], edgecolor='black', alpha=BAR_ALPHA
)
ax_b.set_title('Distribution of Subcellular Localization (Before cutoff)', fontsize=TITLE_SIZE, fontweight='bold')
# ax_b.set_xlabel('Subcellular Localization', fontsize=LABEL_SIZE)
ax_b.set_ylabel('Frequency', fontsize=LABEL_SIZE)
ax_b.set_xticklabels(loc_counts_before.index, rotation=45, ha='right')
beautify_ax(ax_b)
plt.tight_layout()
```

```python
df_filtered = circRNA_df[circRNA_df['RNALocate_Score'] >= 0.5]

# Print the number of rows after filtering.
print(f'Rows after filtering: {len(df_filtered)} rows')

# Plot the filtered distribution.
plt.figure(figsize=(8, 6))
plt.hist(df_filtered['RNALocate_Score'], bins=20, range=(0.5, 1.0), color='skyblue', edgecolor='black')

# Add plot title and axis labels.
plt.title('Distribution of RNALocate_Score (>= 0.5)', fontsize=14)
plt.xlabel('RNALocate_Score', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

# Show the plot.
plt.tight_layout()
plt.show()
```
