```python
import os
import numpy as np
import pandas as pd
import lightgbm as lgb
import shap
import statsmodels.stats.multitest as multi
from scipy.stats import pearsonr
import joblib
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import kruskal
from matplotlib.ticker import MaxNLocator
import random
from Bio import SeqIO
```

```python
# 将负类相关kmer突变为正类相关ker
input_fasta = "/lulabdata3/huangkeyun/zhangys/RNA_locator/ML_python_scripts/reference_preprocessing/linearRNA/linearDataSet/lncRNA_balanced_dataset.fa"
output_fasta = "./lncRNA_EV_gain_mutated.fasta"

EV_kmers = ['TAG', 'TAGA', 'ATTT', 'GAGAG', 'GGGT', 'TTTTA', 'GTA', 'GAG', 'TAA',
       'TTTTT', 'TAGCT', 'ATT']
Cellular_kmers = ['CGCGA', 'CGATA', 'GCGTA', 'ACGCC', 'CGTAG', 'GTTAC', 'GCACG', 'CCGAA',
       'GCGCA', 'CGACC', 'TGCGA', 'GCGCG', 'GTAAG', 'CGTTA', 'CCGGG', 'GTCGT',
       'TATCG', 'TAGCG', 'CATAA', 'TGCGG']

random.seed(198)
# 工具函数：生成非重叠突变坐标
def find_non_overlapping_matches(seq, target_kmers):
    matches = []
    used = set()
    for kmer in target_kmers:
        start = 0
        while True:
            idx = seq.find(kmer, start)
            if idx == -1:
                break
            # 检查是否与已有的匹配区域重叠
            if all((i not in used) for i in range(idx, idx+len(kmer))):
                matches.append((idx, idx+len(kmer), kmer))
                used.update(range(idx, idx+len(kmer)))
            start = idx + 1
    # 按照起点坐标排序
    matches.sort(key=lambda x: x[0])
    return matches

def mutate_sequence(seq_str, target_kmers, replacement_kmers):
    matches = find_non_overlapping_matches(seq_str, target_kmers)
    seq_list = list(seq_str)
    # 反向遍历 matches，避免替换不同长度字符串时导致后续索引偏移
    for start, end, kmer in reversed(matches):
        replacement = random.choice(replacement_kmers)
        # 替换：长度不一致时直接替换start:end为新kmer
        seq_list[start:end] = list(replacement)
    return ''.join(seq_list)

# 读取并突变FASTA
from Bio.Seq import Seq

mutated_records = []
for record in SeqIO.parse(input_fasta, "fasta"):
    seq = str(record.seq)
    
    # 只对序列名以_1结尾的序列进行突变
    if record.id.endswith('_1'):
        # 将 Cellular_kmers 替换为 EV_kmers
        mutated_seq = mutate_sequence(seq, Cellular_kmers, EV_kmers)
        record.seq = Seq(mutated_seq)
    
    mutated_records.append(record)

# 保存突变后的FASTA
with open(output_fasta, "w") as f:
    SeqIO.write(mutated_records, f, "fasta")
```
