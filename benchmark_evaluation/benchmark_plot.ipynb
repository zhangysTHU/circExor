```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
from statsmodels.stats.multitest import multipletests

# 指标名称
metrics = ['AUROC', 'AUPRC', 'Accuracy', 'F1 Score', 'MCC']

# 数据矩阵字典 (模型名称 -> 5个 Bootstrap x 5个 指标)
model_data_dict = {
    'circExor': np.array([
        [0.7050, 0.6650, 0.6550, 0.6683, 0.3094],
        [0.6993, 0.6513, 0.6400, 0.6382, 0.2805],
        [0.6446, 0.5909, 0.6000, 0.5939, 0.2001],
        [0.6786, 0.6542, 0.6375, 0.6523, 0.2750],
        [0.6748, 0.5979, 0.6100, 0.5916, 0.2185]
    ]),
    'RNAlocate': np.array([
        [0.567164, 0.546018, 0.5725, 0.700525, 0.233213],
        [0.556573, 0.517543, 0.5450, 0.678445, 0.214816],
        [0.542825, 0.522580, 0.5350, 0.679310, 0.160591],
        [0.526486, 0.550998, 0.5575, 0.703518, 0.113076],
        [0.526806, 0.486272, 0.4925, 0.641975, 0.074571]
    ]),
    'exoGRU': np.array([
        [0.460509, 0.504242, 0.4625, 0.505747, -0.079071],
        [0.516598, 0.490370, 0.5075, 0.547126,  0.020419],
        [0.478825, 0.499320, 0.4850, 0.529680, -0.030557],
        [0.470647, 0.527084, 0.4900, 0.542601, -0.032102],
        [0.507435, 0.500000, 0.4800, 0.490196, -0.034986]
    ])
}

# 计算 AUROC（第 0 列）的平均值以进行排序
model_auroc_means = {model: np.mean(data[:, 0]) for model, data in model_data_dict.items()}

# 根据 AUROC 的平均值从低到高对模型进行排序
models_names = sorted(model_auroc_means.keys(), key=lambda k: model_auroc_means[k])
print("根据 AUROC (低 -> 高) 的排序结果: ", models_names)

# 组织 sorted_data (模型数, Bootstrap数, 指标数)
n_bootstraps = 5
data_raw = np.array([model_data_dict[m] for m in models_names])
data_mean = np.mean(data_raw, axis=1)
data_std = np.std(data_raw, axis=1, ddof=1)
# 计算 95% 置信区间
t_critical = stats.t.ppf(0.975, n_bootstraps - 1)
data_ci = (data_std / np.sqrt(n_bootstraps)) * t_critical

# 暖色调配色（从浅到深）
warm_colors = ['#F8D7B4', '#F0C9A8', '#E8BB99', '#D89F7B', '#C8835D', '#B05930']
model_colors = [warm_colors[1], warm_colors[3], warm_colors[5]]  # 抽三种从浅到深的颜色

# 设置图形参数
fig, ax = plt.subplots(figsize=(14, 7))
n_metrics = len(metrics)
n_models = len(models_names)
bar_width = 0.8 / (n_models + 0.5)
x = np.arange(n_metrics)

# 记录每个柱子的x坐标，结构为 (n_metrics, n_models)
bar_x_positions = np.zeros((n_metrics, n_models))

min_y = np.min(data_mean - data_ci)

# 绘制分组柱子
for i, model_name in enumerate(models_names):
    offset = (i - n_models/2 + 0.5) * bar_width
    bar_x_positions[:, i] = x + offset
    
    bars = ax.bar(bar_x_positions[:, i], data_mean[i], bar_width, 
                  label=model_name, color=model_colors[i], 
                  edgecolor='black', linewidth=1.2, alpha=0.85,
                  yerr=data_ci[i], capsize=6, error_kw={'linewidth': 1.2, 'capthick': 1.2})
    
    # 在柱子上添加数值标签
    for j, bar in enumerate(bars):
        height = data_mean[i, j]
        if height >= 0:
            y_pos = height + data_ci[i, j] + 0.01
            va = 'bottom'
        else:
            # 强制将负数的数值标签显示在 0 线之上
            y_pos = 0.015
            va = 'bottom'

        ax.text(bar.get_x() + bar.get_width()/2., y_pos,
                f'{height:.3f}', ha='center', va=va, fontsize=10, fontweight='bold')

# ================= 添加显著性检验 =================
def get_asterisks(p):
    if p < 0.001: return '***'
    elif p < 0.01: return '**'
    elif p < 0.05: return '*'
    return 'ns'

best_model_idx = n_models - 1 # 最右侧的模型索引 (AUROC 最高)

for m_idx in range(n_metrics):
    p_values = []
    compare_indices = []
    
    best_raw = data_raw[best_model_idx, :, m_idx]
    
    # 与左侧所有模型进行比较
    for left_idx in range(n_models - 1):
        left_raw = data_raw[left_idx, :, m_idx]
        # 配对单尾 t 检验 (最右侧模型 > 左侧模型)
        _, p_val = stats.ttest_rel(best_raw, left_raw, alternative='greater')
        p_values.append(p_val)
        compare_indices.append(left_idx)
        
    # FDR 多重比较校正
    _, p_adj, _, _ = multipletests(p_values, alpha=0.05, method='fdr_bh')
    
    # 获取该指标下的最大高度 (Mean + CI) 用于标注基础高度
    group_max_y = np.max(data_mean[:, m_idx] + data_ci[:, m_idx])
    base_y_offset = group_max_y + 0.06  # 留出写数值标签的空间
    
    for offset_idx, (left_idx, p_val) in enumerate(zip(compare_indices, p_adj)):
        stars = get_asterisks(p_val)
        if stars != 'ns':
            x1 = bar_x_positions[m_idx, left_idx]
            x2 = bar_x_positions[m_idx, best_model_idx]
            
            # 为了避免线重合，根据跨度调整每一层连线的高度
            y = base_y_offset + (n_models - 2 - offset_idx) * 0.04
            h = 0.015
            
            # 绘制连线
            ax.plot([x1, x1, x2, x2], [y, y+h, y+h, y], lw=1.2, color='black')
            # 绘制星号
            ax.text((x1+x2)/2, y+h, stars, ha='center', va='bottom', color='black', fontsize=12, fontweight='bold')

# 设置图形属性
ax.set_xticks(x)
ax.set_xticklabels(metrics, fontsize=16, fontweight='bold')
ax.tick_params(axis='y', labelsize=14)

# 自适应 Y 轴范围
global_max_y = ax.get_ylim()[1]
ax.set_ylim(min_y * 1.5 if min_y < 0 else 0, global_max_y + 0.1) 

ax.axhline(0, color='black', linewidth=1)
ax.grid(axis='y', alpha=0.3, linestyle='--')

ax.set_ylabel('Score', fontsize=16, fontweight='bold')
# ax.set_title('Models Performance Comparison (5 Bootstraps with 95% CI)', fontsize=18, fontweight='bold', pad=15)

ax.legend(fontsize=14, loc='upper right')

plt.tight_layout()
plt.show()
```
